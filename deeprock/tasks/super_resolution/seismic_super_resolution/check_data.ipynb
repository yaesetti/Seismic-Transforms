```python
import numpy as np

high_res_data_path = "/workspaces/seismic-datasets/data/original/seismicSuperResolutionData/seismicSuperResolutionData/sx/0100.dat"
low_res_data_path = "/workspaces/seismic-datasets/data/original/seismicSuperResolutionData/seismicSuperResolutionData/nx2/0100.dat"

high_res_shape = (256, 256)
low_res_shape = (128, 128)

high_res_data = np.fromfile(high_res_data_path, dtype=np.float32).reshape(high_res_shape).transpose(1, 0)
low_res_data = np.fromfile(low_res_data_path, dtype=np.float32).reshape(low_res_shape).transpose(1, 0)
print("High-resolution data shape:", high_res_data.shape)
print("Low-resolution data shape:", low_res_data.shape)
```

```python
import matplotlib.pyplot as plt
plt.subplot(1, 2, 1)
plt.imshow(high_res_data, cmap='gray')
plt.title('High-Resolution Data')
plt.subplot(1, 2, 2)
plt.imshow(low_res_data, cmap='gray')
plt.title('Low-Resolution Data')
plt.show()
```
